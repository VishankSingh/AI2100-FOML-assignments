<dd> <h6> <center> $ \textbf{y} = h(\textbf{W}_{3}g(\textbf{W}_{2}g(\textbf{W}_{1}\textbf{x}))) $ </center> </dd> </h6>

```python
##################################################
#Load MNIST data.
##################################################
import torchvision.datasets as datasets
mnist_trainset = datasets.MNIST(root='./data', train=True, download=True, transform=None)
mnist_testset = datasets.MNIST(root='./data', train=False, download=True, transform=None)

mnist_traindata = mnist_trainset.data.numpy()
mnist_trainlabel = mnist_trainset.targets.numpy()

mnist_testdata = mnist_testset.data.numpy()
mnist_testlabel = mnist_testset.targets.numpy()

##################################################
#Define the architecture
##################################################

def ReLu(inp):
    outp = np.maximum(0, inp)
    return outp

def gradReLu(inp):
    outp = (inp > 0).astype(float)
    return outp

def softmax(inp):
    exp = np.exp(inp - np.max(inp, axis=1, keepdims=True))
    outp = exp / np.sum(exp, axis=1, keepdims=True)
    return outp

hidden_activation_function = ReLu
output_activation_function = softmax
hidden_activation_function_grad = gradReLu

def fwdPropagate(inputs, weights):
    W1, W2, W3, b1, b2, b3 = weights
    a1 = np.dot(inputs, W1) + b1
    z1 = hidden_activation_function(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = hidden_activation_function(a2)
    a3 = np.dot(z2, W3) + b3
    z3 = output_activation_function(a3)
    outps = {
        'a1': a1,
        'z1': z1,
        'a2': a2,
        'z2': z2,
        'a3': a3,
        'z3': z3,
    }
    return outps


def computeGradients(inputs, targets, weights, activations):
    W1, W2, W3, b1, b2, b3 = weights
    m = inputs.shape[0]
    z1, z2, z3 = activations['z1'], activations['z2'], activations['z3']
    a1, a2, a3 = activations['a1'], activations['a2'], activations['a3']

    Y = np.zeros_like(z3)
    Y[np.arange(m), targets] = 1

    delta3 = z3 - Y
    dj_dw3 = np.dot(z2.T, delta3) / m
    dj_db3 = np.sum(delta3, axis=0, keepdims=True) / m

    delta2 = np.dot(delta3, W3.T) * hidden_activation_function_grad(a2)
    dj_dw2 = np.dot(z1.T, delta2) / m
    dj_db2 = np.sum(delta2, axis=0, keepdims=True) / m

    delta1 = np.dot(delta2, W2.T) * hidden_activation_function_grad(a1)
    dj_dw1 = np.dot(inputs.T, delta1) / m
    dj_db1 = np.sum(delta1, axis=0, keepdims=True) / m

    return [dj_dw1, dj_dw2, dj_dw3, dj_db1, dj_db2, dj_db3]

def applyGradients(weights, gradients, learning_rate):
    W1, W2, W3, b1, b2, b3 = weights
    dj_dw1, dj_dw2, dj_dw3, dj_db1, dj_db2, dj_db3 = gradients

    W1 -= learning_rate * dj_dw1
    W2 -= learning_rate * dj_dw2
    W3 -= learning_rate * dj_dw3
    b1 -= learning_rate * dj_db1
    b2 -= learning_rate * dj_db2
    b3 -= learning_rate * dj_db3

    return [W1, W2, W3, b1, b2, b3]

def backPropagate(inputs, targets, weights, activations, learning_rate):
    gradients = computeGradients(inputs, targets, weights, activations)
    weights = applyGradients(weights, gradients, learning_rate)
    return weights


##################################################
#Train the network
##################################################

def training(inputs, targets_idx, batch_size = 128, epochs=30, train_val_split=0.8, learning_rate=0.001):
    hidden_units = 512
    n_classes =	10
    n_samples, input_dim = inputs.shape

    split = int(n_samples * train_val_split)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    train_idx, val_idx = indices[:split], indices[split:]
    X_train, y_train = inputs[train_idx], targets_idx[train_idx]
    X_val, y_val = inputs[val_idx], targets_idx[val_idx]


    n_batches =	X_train.shape[0] // batch_size

    W1 = np.random.randn(input_dim,   hidden_units) * np.sqrt(2.0/input_dim)
    b1 = np.zeros((1, hidden_units))
    W2 = np.random.randn(hidden_units, hidden_units) * np.sqrt(2.0/hidden_units)
    b2 = np.zeros((1, hidden_units))
    W3 = np.random.randn(hidden_units, n_classes) * np.sqrt(2.0/hidden_units)
    b3 = np.zeros((1, n_classes))
    weights = [W1, W2, W3, b1, b2, b3]


    for epoch in range(epochs):
        perm = np.random.permutation(X_train.shape[0])
        X_train, y_train = X_train[perm], y_train[perm]

        for batch in range(n_batches):
            X_batch = X_train[batch*batch_size:(batch+1)*batch_size]
            Y_batch = y_train[batch*batch_size:(batch+1)*batch_size]

            activations = fwdPropagate(X_batch, weights)
            weights = backPropagate(X_batch, Y_batch, weights, activations, learning_rate)

        train_acts = fwdPropagate(X_train, weights)
        train_preds = np.argmax(train_acts['z3'], axis=1)

        train_acc = np.mean(train_preds == y_train)
        train_loss = np.sum((y_train - train_preds) ** 2) / 2

        val_acts = fwdPropagate(X_val, weights)
        val_preds = np.argmax(val_acts['z3'], axis=1)


        val_acc = np.mean(val_preds == y_val)
        val_loss = np.sum((y_val - val_preds) ** 2) / 2


        print(f"Epoch {epoch+1}/{epochs}, "
                f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_acc:.4f}, "
                f"Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_acc:.4f}")

        # np.savez(f"weights/mnist_dnn_epoch{epoch}.npz", W1=weights[0], W2=weights[1], W3=weights[2])
    
    return weights


X_train_flat = mnist_traindata.reshape(mnist_traindata.shape[0], -1) / 255.0
X_test_flat = mnist_testdata.reshape(mnist_testdata.shape[0], -1) / 255.0

trained_weights = training(X_train_flat, mnist_trainlabel, epochs=10)


##################################################
#Evaluate the performance on test data
##################################################
test_acts = fwdPropagate(X_test_flat, trained_weights)
test_preds = np.argmax(test_acts['z3'], axis=1)
test_acc = np.mean(test_preds == mnist_testlabel)
test_loss = np.sum((mnist_testlabel - test_preds) ** 2) / 2
print(f"Test loss: {test_loss:.4f}, Test acc: {test_acc:.4f}")






```

```python
def Sigmoid(inp):
    outp = 1 / (1 + np.exp(-inp))
    return outp

def gradSigmoid(inp):
    outp = Sigmoid(inp) * (1 - Sigmoid(inp))
    return outp

def Tanh(inp):
    outp = np.tanh(inp)
    return outp

def gradTanh(inp):
    outp = 1 - np.tanh(inp) ** 2
    return outp

def LeakyReLu(inp):
    outp = np.where(inp > 0, inp, 0.01 * inp)
    return outp

def gradLeakyReLu(inp):
    outp = np.where(inp > 0, 1, 0.01)
    return outp
```

```python
##################################################
#Train the network using Adam optimizer
##################################################

hidden_activation_function = LeakyReLu
hidden_activation_function_grad = gradLeakyReLu

def training_adam(inputs, targets_idx,
                  batch_size=128, epochs=30, train_val_split=0.8,
                  lr=0.001, beta1=0.9, beta2=0.999, eps=1e-8):
    hidden_units = 512
    n_classes =	10
    n_samples, input_dim = inputs.shape

    split = int(n_samples * train_val_split)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    train_idx, val_idx = indices[:split], indices[split:]
    X_train, y_train = inputs[train_idx], targets_idx[train_idx]
    X_val, y_val = inputs[val_idx], targets_idx[val_idx]

    n_batches =	X_train.shape[0] // batch_size

    W1 = np.random.randn(input_dim,   hidden_units) * np.sqrt(2.0/input_dim)
    b1 = np.zeros((1, hidden_units))
    W2 = np.random.randn(hidden_units, hidden_units) * np.sqrt(2.0/hidden_units)
    b2 = np.zeros((1, hidden_units))
    W3 = np.random.randn(hidden_units, n_classes)    * np.sqrt(2.0/hidden_units)
    b3 = np.zeros((1, n_classes))
    weights = [W1, W2, W3, b1, b2, b3]

    m, v = [np.zeros_like(W) for W in weights], [np.zeros_like(W) for W in weights]
    t = 0
    n_batches = X_train.shape[0] // batch_size

    for epoch in range(epochs):
        perm = np.random.permutation(X_train.shape[0])
        Xs, ys = X_train[perm], y_train[perm]
        for b in range(n_batches):
            t += 1
            xb = Xs[b*batch_size:(b+1)*batch_size]
            yb = ys[b*batch_size:(b+1)*batch_size]
            acts = fwdPropagate(xb, weights)
            grads = computeGradients(xb, yb, weights, acts)
            for i in range(3):
                m[i] = beta1*m[i] + (1-beta1)*grads[i]
                v[i] = beta2*v[i] + (1-beta2)*(grads[i]**2)
                m_hat = m[i]/(1-beta1**t)
                v_hat = v[i]/(1-beta2**t)
                weights[i] -= lr * m_hat / (np.sqrt(v_hat) + eps)

        train_acts = fwdPropagate(X_train, weights)
        train_preds = np.argmax(train_acts['z3'], axis=1)
        train_acc = np.mean(train_preds == y_train)
        train_loss = np.sum((y_train - train_preds) ** 2) / 2

        val_acts = fwdPropagate(X_val, weights)
        val_preds = np.argmax(val_acts['z3'], axis=1)
        val_acc = np.mean(val_preds == y_val)
        val_loss = np.sum((y_val - val_preds) ** 2) / 2

        print(f"Epoch {epoch+1}/{epochs}, "
                f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_acc:.4f}, "
                f"Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_acc:.4f}")
        
    return weights

def evaluate(weights, inputs, labels):
    test_acts = fwdPropagate(inputs, weights)
    test_preds = np.argmax(test_acts['z3'], axis=1)
    test_acc = np.mean(test_preds == labels)
    test_loss = np.sum((labels - test_preds) ** 2) / 2
    return test_loss, test_acc

X_train_flat = mnist_traindata.reshape(mnist_traindata.shape[0], -1) / 255.0
X_test_flat = mnist_testdata.reshape(mnist_testdata.shape[0], -1) / 255.0

# SGD
w_sgd = training(X_train_flat, mnist_trainlabel, batch_size=128, epochs=10, train_val_split=0.8, learning_rate=0.01)
loss_sgd, acc_sgd = evaluate(w_sgd, X_test_flat, mnist_testlabel)

# Adam
w_adam = training_adam(X_train_flat, mnist_trainlabel, batch_size=128, epochs=10, train_val_split=0.8, lr=0.01, beta1=0.9, beta2=0.999, eps=1e-8)
loss_adam, acc_adam = evaluate(w_adam, X_test_flat, mnist_testlabel)

##################################################
#Compare the accuracies and report your observations
##################################################

print(f"SGD => Test Loss: {loss_sgd:.4f}, Test Acc: {acc_sgd:.4f}")
print(f"Adam => Test Loss: {loss_adam:.4f}, Test Acc: {acc_adam:.4f}")



```

```python
hidden_activation_function = LeakyReLu
output_activation_function = softmax
hidden_activation_function_grad = gradLeakyReLu

##################################################
#Training with weight regularization
##################################################

# region: L2 regularization
def computeGradients(inputs, targets, weights, activations, reg_lambda = 0.0):
    W1, W2, W3, b1, b2, b3 = weights
    m = inputs.shape[0]
    z1, z2, z3 = activations['z1'], activations['z2'], activations['z3']
    a1, a2, a3 = activations['a1'], activations['a2'], activations['a3']
    Y = np.zeros_like(z3)
    Y[np.arange(m), targets] = 1

    delta3 = z3 - Y
    dj_dw3 = np.dot(z2.T, delta3) / m + reg_lambda * W3
    dj_db3 = np.sum(delta3, axis=0, keepdims=True) / m

    delta2 = np.dot(delta3, W3.T) * hidden_activation_function_grad(a2)
    dj_dw2 = np.dot(z1.T, delta2) / m + reg_lambda * W2
    dj_db2 = np.sum(delta2, axis=0, keepdims=True) / m

    delta1 = np.dot(delta2, W2.T) * hidden_activation_function_grad(a1)
    dj_dw1 = np.dot(inputs.T, delta1) / m + reg_lambda * W1
    dj_db1 = np.sum(delta1, axis=0, keepdims=True) / m

    return [dj_dw1, dj_dw2, dj_dw3, dj_db1, dj_db2, dj_db3]

def training(inputs, targets_idx, batch_size = 128, epochs=30, train_val_split=0.8, learning_rate=0.001, reg_lambda=0.01):
    hidden_units = 512
    n_classes =	10
    n_samples, input_dim = inputs.shape

    split = int(n_samples * train_val_split)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    train_idx, val_idx = indices[:split], indices[split:]
    X_train, y_train = inputs[train_idx], targets_idx[train_idx]
    X_val, y_val = inputs[val_idx], targets_idx[val_idx]

    n_batches =	X_train.shape[0] // batch_size
    
    W1 = np.random.randn(input_dim,   hidden_units) * np.sqrt(2.0/input_dim)
    b1 = np.zeros((1, hidden_units))
    W2 = np.random.randn(hidden_units, hidden_units) * np.sqrt(2.0/hidden_units)
    b2 = np.zeros((1, hidden_units))
    W3 = np.random.randn(hidden_units, n_classes) * np.sqrt(2.0/hidden_units)
    b3 = np.zeros((1, n_classes))
    weights = [W1, W2, W3, b1, b2, b3]

    for epoch in range(epochs):
        perm = np.random.permutation(X_train.shape[0])
        X_train, y_train = X_train[perm], y_train[perm]

        for batch in range(n_batches):
            X_batch = X_train[batch*batch_size:(batch+1)*batch_size]
            Y_batch = y_train[batch*batch_size:(batch+1)*batch_size]
            activations = fwdPropagate(X_batch, weights)
            gradients = computeGradients(X_batch, Y_batch, weights, activations, reg_lambda)
            weights = applyGradients(weights, gradients, learning_rate)

        train_acts = fwdPropagate(X_train, weights)
        train_preds = np.argmax(train_acts['z3'], axis=1)
        train_acc = np.mean(train_preds == y_train)
        train_loss = np.sum((y_train - train_preds) ** 2) / 2

        val_acts = fwdPropagate(X_val, weights)
        val_preds = np.argmax(val_acts['z3'], axis=1)
        val_acc = np.mean(val_preds == y_val)
        val_loss = np.sum((y_val - val_preds) ** 2) / 2

        print(f"Epoch {epoch+1}/{epochs}, "
                f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_acc:.4f}, "
                f"Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_acc:.4f}")

        #Save the parameters of network
        # np.savez(f"weights/mnist_dnn_epoch{epoch}.npz", W1=weights[0], W2=weights[1], W3=weights[2])
    
    return weights

# endregion
w_reg = training(X_train_flat, mnist_trainlabel, batch_size=128, epochs=10, train_val_split=0.8, learning_rate=0.01, reg_lambda=1e-4)
loss_reg, acc_reg = evaluate(w_reg, X_test_flat, mnist_testlabel)
print(f"Weight Regularization => Test Loss: {loss_reg:.4f}, Test Acc: {acc_reg:.4f}")

##################################################
#Training with dropout strategy
##################################################

# region: Dropout

def fwdPropagate_dropout(inputs, weights, masks, training=True):
    W1, W2, W3, b1, b2, b3 = weights
    a1 = np.dot(inputs, W1) + b1
    z1 = hidden_activation_function(a1)
    z1 = z1 * masks[0] if training else z1
    a2 = np.dot(z1, W2) + b2
    z2 = hidden_activation_function(a2)
    z2 = z2 * masks[1] if training else z2
    a3 = np.dot(z2, W3) + b3
    z3 = output_activation_function(a3)
    # z3 = z3 * masks[2] if training else z3
    outps = {'a1': a1, 'z1': z1, 'a2': a2, 'z2': z2, 'a3': a3, 'z3': z3}
    return outps

def computeGradients_dropout(inputs, targets, weights, activations, masks):
    W1, W2, W3, b1, b2, b3 = weights
    m = inputs.shape[0]
    z1, z2, z3 = activations['z1'], activations['z2'], activations['z3']
    a1, a2, a3 = activations['a1'], activations['a2'], activations['a3']

    Y = np.zeros_like(z3)
    Y[np.arange(m), targets] = 1

    delta3 = z3 - Y
    # delta3 = delta3 * masks[2] if training else delta3
    dj_dw3 = np.dot(z2.T, delta3) / m
    dj_db3 = np.sum(delta3, axis=0, keepdims=True) / m

    delta2 = np.dot(delta3, W3.T) * hidden_activation_function_grad(a2)
    delta2 = delta2 * masks[1] if training else delta2
    dj_dw2 = np.dot(z1.T, delta2) / m
    dj_db2 = np.sum(delta2, axis=0, keepdims=True) / m

    delta1 = np.dot(delta2, W2.T) * hidden_activation_function_grad(a1)
    delta1 = delta1 * masks[0] if training else delta1
    dj_dw1 = np.dot(inputs.T, delta1) / m
    dj_db1 = np.sum(delta1, axis=0, keepdims=True) / m

    return [dj_dw1, dj_dw2, dj_dw3, dj_db1, dj_db2, dj_db3]

def forward_back_propagate_dropout(inputs, targets, weights, learning_rate, dropout_rate=0.2):
    layer_sizes = [w.shape[1] for w in weights[:3]]  # Get number of output units for each layer
    keep = 1.0 - dropout_rate
    masks = [
      (np.random.binomial(1, keep, size=(1, s)) / keep)
      for s in layer_sizes
    ]

    activations = fwdPropagate_dropout(inputs, weights, masks, training=True)
    gradients = computeGradients_dropout(inputs, targets, weights, activations, masks)
    weights = applyGradients(weights, gradients, learning_rate)
    return weights

def training(inputs, targets_idx, batch_size = 128, epochs=30, train_val_split=0.8, learning_rate=0.001, dropout_rate=0.2):
    hidden_units = 512
    n_classes =	10
    n_samples, input_dim = inputs.shape

    split = int(n_samples * train_val_split)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    train_idx, val_idx = indices[:split], indices[split:]
    X_train, y_train = inputs[train_idx], targets_idx[train_idx]
    X_val, y_val = inputs[val_idx], targets_idx[val_idx]

    n_batches =	X_train.shape[0] // batch_size

    W1 = np.random.randn(input_dim,   hidden_units) * np.sqrt(2.0/input_dim)
    b1 = np.zeros((1, hidden_units))
    W2 = np.random.randn(hidden_units, hidden_units) * np.sqrt(2.0/hidden_units)
    b2 = np.zeros((1, hidden_units))
    W3 = np.random.randn(hidden_units, n_classes) * np.sqrt(2.0/hidden_units)
    b3 = np.zeros((1, n_classes))
    weights = [W1, W2, W3, b1, b2, b3]


    for epoch in range(epochs):
        perm = np.random.permutation(X_train.shape[0])
        X_train, y_train = X_train[perm], y_train[perm]
        for batch in range(n_batches):
            start = batch * batch_size
            end = start + batch_size
            X_batch = X_train[start:end]
            Y_batch = y_train[start:end]
            weights = forward_back_propagate_dropout(X_batch, Y_batch, weights, learning_rate, dropout_rate)

        train_acts = fwdPropagate_dropout(X_train, weights, masks=None, training=False)
        train_preds = np.argmax(train_acts['z3'], axis=1)
        train_acc = np.mean(train_preds == y_train)
        train_loss = np.sum((y_train - train_preds) ** 2) / 2
        val_acts = fwdPropagate_dropout(X_val, weights, masks=None, training=False)
        val_preds = np.argmax(val_acts['z3'], axis=1)
        val_acc = np.mean(val_preds == y_val)
        val_loss = np.sum((y_val - val_preds) ** 2) / 2

        print(f"Epoch {epoch+1}/{epochs}, "
                f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_acc:.4f}, "
                f"Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_acc:.4f}")

    return weights


# endregion
w_reg = training(X_train_flat, mnist_trainlabel, batch_size=128, epochs=10, train_val_split=0.8, learning_rate=0.1)
loss_reg, acc_reg = evaluate(w_reg, X_test_flat, mnist_testlabel)
print(f"Dropout => Test Loss: {loss_reg:.4f}, Test Acc: {acc_reg:.4f}")

##################################################
#Training with early stopping criterion
##################################################

# region: Early stopping

def forward_back_propagate(inputs, targets, weights, learning_rate):
    activations = fwdPropagate(inputs, weights)
    gradients = computeGradients(inputs, targets, weights, activations)
    weights = applyGradients(weights, gradients, learning_rate)
    return weights

def training(inputs, targets_idx, batch_size=128, epochs=30, train_val_split=0.8,
             learning_rate=0.001, patience=5):
    hidden_units = 512
    n_classes = 10
    n_samples, input_dim = inputs.shape

    split = int(n_samples * train_val_split)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    train_idx, val_idx = indices[:split], indices[split:]
    X_train, y_train = inputs[train_idx], targets_idx[train_idx]
    X_val, y_val = inputs[val_idx], targets_idx[val_idx]
    n_batches = X_train.shape[0] // batch_size

    W1 = np.random.randn(input_dim,   hidden_units) * np.sqrt(2.0/input_dim)
    b1 = np.zeros((1, hidden_units))
    W2 = np.random.randn(hidden_units, hidden_units) * np.sqrt(2.0/hidden_units)
    b2 = np.zeros((1, hidden_units))
    W3 = np.random.randn(hidden_units, n_classes)    * np.sqrt(2.0/hidden_units)
    b3 = np.zeros((1, n_classes))
    weights = [W1, W2, W3, b1, b2, b3]


    best_val_loss = float('inf')
    best_weights = None
    no_improve_count = 0

    for epoch in range(epochs):
        perm = np.random.permutation(X_train.shape[0])
        X_train, y_train = X_train[perm], y_train[perm]
        for batch in range(n_batches):
            start, end = batch * batch_size, (batch + 1) * batch_size
            X_batch, Y_batch = X_train[start:end], y_train[start:end]
            weights = forward_back_propagate(X_batch, Y_batch, weights, learning_rate)

        train_preds = np.argmax(fwdPropagate(X_train, weights)['z3'], axis=1)
        val_preds = np.argmax(fwdPropagate(X_val, weights)['z3'], axis=1)
        train_acc = np.mean(train_preds == y_train)
        val_acc = np.mean(val_preds == y_val)
        train_loss = np.sum((train_preds != y_train).astype(np.float32))
        val_loss = np.sum((val_preds != y_val).astype(np.float32))

        print(f"Epoch {epoch+1}/{epochs}, "
              f"Train Loss: {train_loss:.2f}, Train Accuracy: {train_acc:.4f}, "
              f"Validation Loss: {val_loss:.2f}, Validation Accuracy: {val_acc:.4f}")

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_weights = [w.copy() for w in weights]
            no_improve_count = 0
        else:
            no_improve_count += 1
            if no_improve_count >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                break

    return best_weights



# endregion
w_best = training(X_train_flat, mnist_trainlabel, batch_size=128, epochs=10, learning_rate=0.1)
loss_final, acc_final = evaluate(w_best, X_test_flat, mnist_testlabel)
print(f"Early Stopping => Test Loss: {loss_final:.4f}, Test Acc: {acc_final:.4f}")
```

```python
# logistic regression code
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

def Logistic_Regression(X_train, Y_train, X_test):
  #Inputs: Training data, Training labels, and Testing data
  #Outputs: Predictions on test data

  max_iter = 1000  #max number of ierations for parameter update
  threshold = 0.5 #threshold for classification
  X = np.hstack((np.ones((X_train.shape[0], 1)), X_train)) #Append ones to the training data to account for the bias term
  w = np.zeros(X.shape[1]) #Initialize the weight vector to zeros
  for iter in range(0,max_iter,1):
    preds = sigmoid(X @ w)
    R = np.diag(preds * (1 - preds)) 
    error = preds - Y_train

    # w_old = w
    # print((np.linalg.pinv(X.T @ R @ X) @ X.T @ R @ (X @ w_old - np.linalg.inv(R) @ error)).shape)
    # w = np.linalg.pinv(X.T @ R @ X) @ X.T @ R @ (X @ w_old - np.linalg.inv(R) @ error)

    hessian = X.T @ R @ X + 1e-6 * np.eye(X.shape[1])

    gradient = X.T @ error
      
    delta = np.linalg.pinv(hessian) @ gradient
    w -= delta
    if np.linalg.norm(delta) < 1e-6: 
      break


  #Test data predictions
  X_test = np.hstack((np.ones((X_test.shape[0], 1)), X_test))
  test_preds = sigmoid(X_test @ w)
  test_preds = (test_preds >= threshold).astype(int)

  #Return the predicted test labels
  return test_preds, w

def LR_Accuracy(Y_test, Y_pred):
  #Inputs: Ground truth and predicted lables of test data
  #Outputs: Accuracy
  acc = np.mean(Y_test == Y_pred, axis=0)

  #Compute and return the accuracy
  return acc

```

```python
##################################################
#Classify the linearly separable data using linear classification models
##################################################


Y_pred, w_logistic = Logistic_Regression(tr_data, tr_targets, te_data)
acc = LR_Accuracy(te_targets, Y_pred)
print('Accuracy with Logistic Regression:', acc)

# region: Plot the decision boundary
x = np.linspace(-1, 4, 100)

def predict_w3_zero_one(x, weight) -> np.float64:
    return -(weight[1]*x + weight[0] - 0.5)/weight[2]

plt.plot(x, predict_w3_zero_one(x, w_logistic.T), color='black', label = "decision boundary")
plt.scatter(class1_te[:, 0], class1_te[:, 1], color='cyan', label = "class 1", s=10)
plt.scatter(class2_te[:, 0], class2_te[:, 1], color='pink', label = "class 2", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("Logistic Regression")
plt.grid()
plt.legend()
plt.show()

# endregion


##################################################
#Classify the linearly separable data using deep neural based classification models
##################################################

# region: dnn
def relu(x):
    return np.maximum(0, x)

def relu_deriv(x):
    return (x > 0).astype(float)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def loss_function(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-10, 1 - 1e-10)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

def LR_Accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)

class DNN:
    def __init__(self, input_dim=2, hidden1=3, hidden2=3, output_dim=1):
        self.W1 = np.random.randn(hidden1, input_dim) * 0.1
        self.b1 = np.zeros((hidden1, 1))
        self.W2 = np.random.randn(hidden2, hidden1) * 0.1
        self.b2 = np.zeros((hidden2, 1))
        self.W3 = np.random.randn(output_dim, hidden2) * 0.1
        self.b3 = np.zeros((output_dim, 1))

    def forward(self, x):
        self.x = x.T
        self.z1 = self.W1 @ self.x + self.b1
        self.a1 = relu(self.z1)
        self.z2 = self.W2 @ self.a1 + self.b2
        self.a2 = relu(self.z2)
        self.z3 = self.W3 @ self.a2 + self.b3
        self.a3 = sigmoid(self.z3)
        return self.a3.T.squeeze()

    def train(self, X, y, lr=0.01, epochs=5000):
        X = X.T
        y = y.reshape(1, -1)
        for epoch in range(epochs):
            z1 = self.W1 @ X + self.b1
            a1 = relu(z1)
            z2 = self.W2 @ a1 + self.b2
            a2 = relu(z2)
            z3 = self.W3 @ a2 + self.b3
            a3 = sigmoid(z3)

            loss = loss_function(y, a3)
            # if epoch % 1000 == 0:
                # print(f"Epoch {epoch}, Loss: {loss:.4f}")

            dz3 = a3 - y
            dW3 = dz3 @ a2.T / X.shape[1]
            db3 = np.mean(dz3, axis=1, keepdims=True)

            dz2 = (self.W3.T @ dz3) * relu_deriv(z2)
            dW2 = dz2 @ a1.T / X.shape[1]
            db2 = np.mean(dz2, axis=1, keepdims=True)

            dz1 = (self.W2.T @ dz2) * relu_deriv(z1)
            dW1 = dz1 @ X.T / X.shape[1]
            db1 = np.mean(dz1, axis=1, keepdims=True)

            self.W3 -= lr * dW3
            self.b3 -= lr * db3
            self.W2 -= lr * dW2
            self.b2 -= lr * db2
            self.W1 -= lr * dW1
            self.b1 -= lr * db1

    def predict(self, X):
        return self.forward(X) > 0.5
    
    def get_second_layer_activations(self, X):
        X = X.T
        z1 = self.W1 @ X + self.b1
        a1 = relu(z1)
        z2 = self.W2 @ a1 + self.b2
        a2 = relu(z2)
        return a2.T  # Shape: (n_samples, 3)


nn = DNN()
nn.train(tr_data, tr_targets, lr=0.01, epochs=10000)
pred_nn = nn.predict(te_data)
acc_nn = LR_Accuracy(te_targets, pred_nn)
print(f"DNN Accuracy: {acc_nn:.4f}")

x_vals = np.linspace(-2, 6, 300)
y_vals = np.linspace(-2, 6, 300)
xx, yy = np.meshgrid(x_vals, y_vals)
grid = np.c_[xx.ravel(), yy.ravel()]

probs = nn.forward(grid)
Z = probs.reshape(xx.shape)

plt.contourf(xx, yy, Z, levels=[0, 0.5, 1], colors=["#99ccff", "#ffcccc"], alpha=0.4)
plt.contour(xx, yy, Z, levels=[0.5], colors='black', linewidths=2)

plt.scatter(class1_te[:, 0], class1_te[:, 1], color='cyan', label="class 2", s=10)
plt.scatter(class2_te[:, 0], class2_te[:, 1], color='pink', label="class 1", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("Neural Network Decision Boundary")
plt.grid(True)
plt.legend()
plt.show()


a2_activations = nn.get_second_layer_activations(tr_data)

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

colors = ['red' if label == 0 else 'blue' for label in tr_targets]

ax.scatter(a2_activations[:, 0], a2_activations[:, 1], a2_activations[:, 2],
           c=colors, alpha=0.7, s=20)

ax.set_xlabel("Activation 1")
ax.set_ylabel("Activation 2")
ax.set_zlabel("Activation 3")
ax.set_title("3D Second-Layer Activations of DNN")
plt.show()


##################################################
#Compare the performance of both the models on linearly separable data
##################################################








```
