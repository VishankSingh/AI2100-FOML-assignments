```python
########################################
#Part-2)-(a): Data generation
########################################

# region 

#Set the parameters of class distribution
mean1 = [1,1]           #Mean of class1
mean2 = [2,2]           #Mean of class2
cov = [[0.3,0],[0,0.3]] #Covariance matrix. Same covariance for class1 and class2

np.random.seed(14)

label1 = 0
label2 = 1

#Generate training data. You can use "np.random.multivariate_normal()" function to sample data points from multivariate Gaussian distribution
class1_tr = np.random.multivariate_normal(mean1, cov, 50)
class2_tr = np.random.multivariate_normal(mean2, cov, 50)
tr_data = np.vstack((class1_tr, class2_tr))
tr_targets = np.array([label1] * 50 + [label2] * 50)

#Generate testing data.
class1_te = np.random.multivariate_normal(mean1, cov, 200)
class2_te = np.random.multivariate_normal(mean2, cov, 200)
te_data = np.vstack((class1_te, class2_te))
te_targets = np.array([label1] * 200 + [label2] * 200)

# endregion


########################################
#Part-2)-(b)
########################################

# region: LDA
#Complete the below function
def LDA_classify(X_train, Y_train, X_test):
  #Inputs: Training data, Training labels, and Testing data
  #Outputs: Between class covariance

  class1 = X_train[Y_train == 0]
  class2 = X_train[Y_train == 1]
  
  mean_class1 = np.mean(class1, axis=0)
  mean_class2 = np.mean(class2, axis=0)
  
  w = mean_class2 - mean_class1   
  
  threshold = w @ (mean_class1 + mean_class2) / 2
  projections = X_test @ w
  predictions = (projections >= threshold).astype(int)
  
  return predictions.reshape(-1, 1), w.reshape(-1, 1)

predictions, w = LDA_classify(tr_data, tr_targets, te_data)

# print(predictions.shape)
# print(w.shape)


def LDA_project(w, X_test, Y_test):
  #Inputs: weight vector, testing data, and testing labels

  class1 = X_test[Y_test == 0]
  class2 = X_test[Y_test == 1]
  
  w_unit = w / np.linalg.norm(w)
  w_perp = np.array([-w_unit[1], w_unit[0]])
  w_perp = w_perp / np.linalg.norm(w_perp)

  proj_class1 = w_unit.T * (class1 @ w_unit) 
  proj_class2 = w_unit.T * (class2 @ w_unit) 

  shift_amount = -3  
  proj_class1 = proj_class1 + shift_amount * w_perp.T
  proj_class2 = proj_class2 + (shift_amount) * w_perp.T
  
  plt.figure(figsize=(6, 6))
  plt.scatter(class1[:, 0], class1[:, 1], color='cyan', label="Original Class 1", s=10)
  plt.scatter(class2[:, 0], class2[:, 1], color='magenta', label="Original Class 2", s=10)
  
  plt.scatter(proj_class1[:, 0], proj_class1[:, 1], color='blue', marker='x', label="Projected Class 1", s=10)
  plt.scatter(proj_class2[:, 0], proj_class2[:, 1], color='red', marker='x', label="Projected Class 2", s=10)
  
  for _ in range(class1.shape[0]):
      plt.plot([class1[_, 0], proj_class1[_, 0]], 
               [class1[_, 1], proj_class1[_, 1]], 
               'cyan', alpha=0.1, lw=0.8, ls='--')
  for _ in range(class2.shape[0]):
      plt.plot([class2[_, 0], proj_class2[_, 0]], 
               [class2[_, 1], proj_class2[_, 1]], 
               'magenta', alpha=0.1, lw=0.8, ls='--')
  
  overall_mean = np.mean(X_test, axis=0)
  line_points = np.array([overall_mean - 5 * w_unit.flatten(), overall_mean + 5 * w_unit.flatten()])
  plt.plot(line_points[:, 0], line_points[:, 1], 'k--', label="Projection Direction")
  
  plt.xlabel("Coordinate 1")
  plt.ylabel("Coordinate 2")
  plt.title("2-D Data and Their 1-D Projections (Between-Class Variance Only)")
  plt.legend()
  plt.grid()
  plt.xlim(-1, 5)
  plt.ylim(-2, 4)
  plt.show()
  
  projections = (X_test @ w_unit).flatten()
  plt.figure()
  plt.hist(projections[Y_test == 0], bins=30, alpha=0.6, label="Class 1")
  plt.hist(projections[Y_test == 1], bins=30, alpha=0.6, label="Class 2")
  plt.xlabel("Projection Value")
  plt.ylabel("Frequency")
  plt.title("Histogram of 1-D Projections")
  plt.grid(True)
  plt.show()


LDA_project(w, te_data, te_targets)
#Compute and print the accuracy
acc = LS_Classify_Accuracy(te_targets, predictions.flatten())
print('Accuracy with LDA:', acc*100)
# endregion

########################################
#Part-2)-(c)
########################################

# region: FDA
def FDA_classify(X_train, Y_train, X_test):
  #Inputs: Training data, Training labels, and Testing data
  #Outputs: Weight vector and predictions

  class1 = X_train[Y_train == 0]
  class2 = X_train[Y_train == 1]
  
  mean_class1 = np.mean(class1, axis=0)
  mean_class2 = np.mean(class2, axis=0)
  
  S1 = np.cov(class1, rowvar=False)
  S2 = np.cov(class2, rowvar=False)
  
  Sw = S1 + S2
  
  # Compute Fisher's discriminant vector: w = Sw^{-1}(mean2 - mean1)
  w = np.linalg.inv(Sw) @ (mean_class2 - mean_class1).T
  
  # Compute the threshold as the average of the projected class means
  threshold = w @ (mean_class1 + mean_class2) / 2
  
  # Project the test data onto w
  projections = X_test @ w
  
  # Classify based on the threshold
  predictions = (projections >= threshold).astype(int)
  
  return predictions.reshape(-1, 1), w.reshape(-1, 1)


predictions, w = FDA_classify(tr_data, tr_targets, te_data)


def FDA_project(w, X_test, Y_test):
  #Inputs: weight vector, testing data, and testing labels

  #Project the test data onto 1-dimensional space
  class1 = X_test[Y_test == 0]
  class2 = X_test[Y_test == 1]
  
  # Normalize the weight vector to obtain the projection direction
  w_unit = w / np.linalg.norm(w)
  w_perp = np.array([-w_unit[1], w_unit[0]])
  w_perp = w_perp / np.linalg.norm(w_perp)

  
  proj_class1 = (class1 @ w_unit) * w_unit.T
  proj_class2 = (class2 @ w_unit) * w_unit.T

  shift_amount = -3  
  proj_class1 = proj_class1 + shift_amount * w_perp.T
  proj_class2 = proj_class2 + (shift_amount) * w_perp.T
  
  # Plot the original data points and the projected points.
  plt.figure(figsize=(6, 6))
  plt.scatter(class1[:, 0], class1[:, 1], color='cyan', label="Original class 1", s=10)
  plt.scatter(class2[:, 0], class2[:, 1], color='magenta', label="Original class 2", s=10)
  
  # Plot the projected points (using marker 'x')
  plt.scatter(proj_class1[:, 0], proj_class1[:, 1], color='blue', alpha=1 ,marker='x', label="Projected class 1", s=5)
  plt.scatter(proj_class2[:, 0], proj_class2[:, 1], color='red', alpha=1 ,marker='x', label="Projected class 2", s=5)
  
  # Draw dashed, thin lines connecting each original point with its projection.
  for i in range(class1.shape[0]):
      plt.plot([class1[i, 0], proj_class1[i, 0]], 
               [class1[i, 1], proj_class1[i, 1]], 
               color='cyan', alpha=0.1, lw=0.8, ls='--')
  for i in range(class2.shape[0]):
      plt.plot([class2[i, 0], proj_class2[i, 0]], 
               [class2[i, 1], proj_class2[i, 1]], 
               color='magenta', alpha=0.1, lw=0.8, ls='--')
  
  # Optionally, plot the projection direction line.
  # Here, we plot a line through the overall mean along the w_unit direction.
  overall_mean = np.mean(X_test, axis=0)
  line_points = np.array([overall_mean - 5 * w_unit.flatten(), overall_mean + 5 * w_unit.flatten()])
  # line_points = line_points + shift_amount * w_perp.T
  plt.plot(line_points[:, 0], line_points[:, 1], 'k--', lw=1, label="Projection Direction")
  
  plt.xlabel("Coordinate 1")
  plt.ylabel("Coordinate 2")
  plt.title("2-D Data and Their 1-D Projections (Fisher Discriminant)")
  plt.legend(loc='upper right')
  plt.grid()
  plt.xlim(-1, 5)
  plt.ylim(-2, 4)
  plt.show()
  
  # Also plot a histogram of the 1-D projections.
  projections = X_test.dot(w_unit).flatten()
  plt.figure()
  plt.hist(projections[Y_test == 0], bins=30, alpha=0.6, label="Class 1")
  plt.hist(projections[Y_test == 1], bins=30, alpha=0.6, label="Class 2")
  plt.xlabel("Projection Value")
  plt.ylabel("Frequency")
  plt.title("Histogram of 1-D Projections")
  plt.grid(True)
  plt.legend()
  plt.show()


  #Plot the projections
  return


FDA_project(w, te_data, te_targets)
#Compute and print the accuracy
acc = LS_Classify_Accuracy(te_targets, predictions.flatten())
print('Accuracy with FDA:', acc*100)
# endregion

########################################
#Part-2)-(d)
########################################

# region: FDA_project_K
def FDA_project_K(X_train, Y_train, X_test, k):
  #Inputs: Training data, Training label, Testing data, and the number of eigen vectors to be considered
  #Oupur: Projection matrix
  Y_train = Y_train.flatten()
  
  classes = np.unique(Y_train)
  n_classes = len(classes)
  n_features = X_train.shape[1]
  
  overall_mean = np.mean(X_train, axis=0)
  
  S_W = np.zeros((n_features, n_features))
  S_B = np.zeros((n_features, n_features))
  
  for c in classes:
      X_c = X_train[Y_train == c]
      n_c = X_c.shape[0]
      mean_c = np.mean(X_c, axis=0)
      S_W += np.cov(X_c, rowvar=False) * (n_c - 1)

      mean_diff = (mean_c - overall_mean).reshape(-1, 1)
      S_B += n_c * (mean_diff @ mean_diff.T)
  
  eigvals, eigvecs = np.linalg.eig(np.linalg.pinv(S_W) @ S_B)
  
  sorted_indices = np.argsort(eigvals.real)[::-1]
  eigvecs = eigvecs[:, sorted_indices]
  
  W = eigvecs[:, :k].real
  
  X_proj = X_test @ W
  
  #Compute and return the projection matrix
  return W, X_proj


np.random.seed(14)
n_samples_per_class = 50
n_features = 3

mean1 = [1, 1, 1]
mean2 = [3, 3, 3]
mean3 = [5, 5, 5]
cov = np.eye(n_features) * 0.5

X1 = np.random.multivariate_normal(mean1, cov, n_samples_per_class)
X2 = np.random.multivariate_normal(mean2, cov, n_samples_per_class)
X3 = np.random.multivariate_normal(mean3, cov, n_samples_per_class)

X = np.vstack((X1, X2, X3))
Y = np.array([0]*n_samples_per_class + [1]*n_samples_per_class + [2]*n_samples_per_class)

K = 2
W, X_proj = FDA_project_K(X, Y, X, K)

print("\n\n\n\nOriginal data shape:", X.shape)
print("Projection matrix W (shape):", W.shape)
print("Projected data (shape):", X_proj.shape)

# endregion

```

```python
########################################
#Part-3)-(a): Data generation
########################################

# region: without ouliers

#Set the parameters of class distribution
mean1 = [1,1]           #Mean of class1
mean2 = [2,2]           #Mean of class2
cov = [[0.3,0],[0,0.3]] #Covariance matrix. Same covariance for class1 and class2

np.random.seed(14)

label1 = 0
label2 = 1

#Generate training data. You can use "np.random.multivariate_normal()" function to sample data points from multivariate Gaussian distribution
class1_tr = np.random.multivariate_normal(mean1, cov, 50)                #Class1 training data
class2_tr = np.random.multivariate_normal(mean2, cov, 50)                #Class2 training data
tr_data = np.vstack((class1_tr, class2_tr))              #Combine class1 training and class2 training data to get total training data
tr_targets = np.array([label1] * 50 + [label2] * 50)           #Class labels. Assign the value "0" to class1 and the value "1" to class2.

#Generate testing data.
class1_te = np.random.multivariate_normal(mean1, cov, 200)            #Class1 testing data
class2_te = np.random.multivariate_normal(mean2, cov, 200)            #Class2 testing data
te_data = np.vstack((class1_te, class2_te))              #Combine class1 testing and class2 testing data to get total testing data
te_targets = np.array([label1] * 200 + [label2] * 200)           #Class labels. Assign the value "0" to class1 and the value "1" to class2.


########################################
#Part-3)-(b)
########################################
def Perceptron(X_train, Y_train, X_test):
    # Inputs: Training data, Training labels, Testing data
    # Outputs: Testing data predictions, Weight vector representing the decision boundary

    epochs = 100  # Number of epochs
    # Append ones to the training data to account for the bias term
    X = np.hstack((X_train, np.ones((X_train.shape[0], 1))))
    # Initialize the weight vector (including bias) to zeros
    w = np.zeros(X.shape[1])
    
    # Learning rate (can be tuned)
    lr = 1.0
    
    for epoch in range(epochs):
        # Compute predictions for each sample
        predictions = (X @ w >= 0).astype(int)
        # Update the weights for misclassified samples
        for i in range(X.shape[0]):
            error = Y_train[i] - predictions[i]
            if error != 0:
                # Update rule: w = w + lr * error * x_i
                w = w + lr * error * X[i]
                
    # Compute the test data predictions using the final weight vector
    X_test_bias = np.hstack((X_test, np.ones((X_test.shape[0], 1))))
    predictions = (X_test_bias @ w >= 0.5).astype(int)
    
    return predictions, w


##########################################
#Plot the decision boundary using perceptron
##########################################
def plot_decision_regions(X, y, w, resolution=0.02, title="Perceptron Decision Regions", targets=[0, 1]):

    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                           np.arange(x2_min, x2_max, resolution))

    grid = np.array([xx1.flatten(), xx2.flatten()]).T
    grid_bias = np.hstack((grid, np.ones((grid.shape[0], 1))))

    Z_decision = (grid_bias @ w).reshape(xx1.shape)
    
    # Plot only the decision boundary line (where the decision function equals 0)
    plt.contour(xx1, xx2, Z_decision, levels=[0], colors='black', linewidths=1)

    # Z = (grid_bias @ w >= 0).astype(int)
    # Z = Z.reshape(xx1.shape)

    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())
    

    plt.scatter(X[y == targets[0], 0], X[y == targets[0], 1], color='cyan', label="Class 0", s=10)
    plt.scatter(X[y == targets[1], 0], X[y == targets[1], 1], color='pink', label="Class 1", s=10)

    plt.xlabel("Coordinate 1")
    plt.ylabel("Coordinate 2")
    plt.title(title)
    plt.legend()
    plt.grid()
    plt.show()


# Train the perceptron and get predictions on test data
predictions, w = Perceptron(tr_data, tr_targets, te_data)

# Print accuracy on test data
accuracy = np.mean(predictions == te_targets)
print("Accuracy with Perceptron:", accuracy)

plot_decision_regions(te_data, te_targets, w)


##########################################
#Plot the decision boundary using least squares
##########################################

def LS_Classify(X_train, Y_train, X_test) -> tuple[np.ndarray, np.ndarray]:
    tr_input = np.ones((X_train.shape[0], 1))
    tr_input = np.hstack((tr_input, X_train))
    tr_target = Y_train
    w_opt = np.linalg.pinv(tr_input.T @ tr_input) @ tr_input.T @ tr_target
    w_opt = np.array(w_opt)
    te_input = np.ones((X_test.shape[0], 1))
    te_input = np.hstack((te_input, X_test))
    pred = te_input @ w_opt
    pred = (pred >= 0.5).astype(int)
    pred = np.array(pred)
    return pred, w_opt

predictions, w_opt = LS_Classify(tr_data, tr_targets, te_data)
# print("Weight vector:", w_opt)

# Print accuracy on test data
accuracy = np.mean(predictions == te_targets)
print("Accuracy with Least Squares:", accuracy)

def predict_w3_zero_one(x, weight) -> np.float64:
    return -(weight[1]*x + weight[0] - 0.5)/weight[2]

x = np.linspace(-1.5, 4.5, 100)

plt.plot(x, predict_w3_zero_one(x, w_opt), color='black', label = "Decision Boundary", lw=1)
plt.scatter(class1_te[:, 0], class1_te[:, 1], color='cyan', label = "True Class 1", s=10)
plt.scatter(class2_te[:, 0], class2_te[:, 1], color='pink', label = "True Class 2", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("LS classification for labels 0/1 (Gaussian Data)")
plt.grid()
plt.legend()
plt.xlim(-1.5, 4.5)
plt.ylim(-1.5, 4.5)
plt.show()

# endregion


##########################################
#Part-3)-(c): Repeat the experiments with the true outliers added to the data. Compare and contrast the perceptron algorithm with least squares method
##########################################

# region: with outliers

# region: Data generation
mean1 = [1,1]
mean2 = [2,2]
cov = [[0.3,0],[0,0.3]]
mean_outlier = [20, 20]

np.random.seed(14)

label1 = 0
label2 = 1

class1_tr = np.random.multivariate_normal(mean1, cov, 50)
class2_tr = np.random.multivariate_normal(mean2, cov, 50)
class2_tr_outlier = np.random.multivariate_normal(mean_outlier, cov, 20)
tr_data = np.vstack((class1_tr, class2_tr, class2_tr_outlier))
tr_targets = np.array([label1] * 50 + [label2] * 50 + [label2] * 20)

class1_te = np.random.multivariate_normal(mean1, cov, 200)
class2_te = np.random.multivariate_normal(mean2, cov, 200)
te_data = np.vstack((class1_te, class2_te))
te_targets = np.array([label1] * 200 + [label2] * 200)

plt.scatter(tr_data[tr_targets == label1, 0], tr_data[tr_targets == label1, 1], color='cyan', label="Class 0", s=10)
plt.scatter(tr_data[tr_targets == label2, 0], tr_data[tr_targets == label2, 1], color='pink', label="Class 1", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("Train Data with Outliers")
plt.legend()
plt.grid()
plt.show()

# endregion

# region: Perceptron plot outliers

# Train the perceptron and get predictions on test data
predictions, w = Perceptron(tr_data, tr_targets, te_data)

# Print accuracy on test data
accuracy = np.mean(predictions == te_targets)
print("Accuracy with Perceptron:", accuracy)

plot_decision_regions(te_data, te_targets, w, title="Perceptron Decision Regions with Outliers")

# endregion

# region: Least Squares plot outliers

# Train the least squares classifier and get predictions on test data
predictions, w_opt = LS_Classify(tr_data, tr_targets, te_data)

accuracy = np.mean(predictions == te_targets)
print("Accuracy with Least Squares:", accuracy)

def predict_w3_zero_one(x, weight) -> np.float64:
    return -(weight[1]*x + weight[0] - 0.5)/weight[2]

x = np.linspace(-1.5, 4.5, 100)

plt.plot(x, predict_w3_zero_one(x, w_opt), color='black', label = "Decision Boundary", lw=1)
plt.scatter(class1_te[:, 0], class1_te[:, 1], color='cyan', label = "True Class 1", s=10)
plt.scatter(class2_te[:, 0], class2_te[:, 1], color='pink', label = "True Class 2", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("LS classification for labels 0/1 (Gaussian Data) with Outliers")
plt.grid()
plt.legend()
plt.xlim(-1.5, 4.5)
plt.ylim(-1.5, 4.5)
plt.show()

# endregion

# endregion


##########################################
#Part-3)-(d): Data generation XOR
##########################################

# region: part 3 d

# region: Data generation XOR

#Set the parameters of Gaussian distribution
class1_mean1 = [0, 0]
class1_mean2 = [0 ,1]
class2_mean1 = [1, 0]
class2_mean2 = [1, 1]
cov = [[0.3,0],[0,0.3]]

# #Generate  data. You can use "np.random.multivariate_normal()" function to sample data points from multivariate Gaussian distribution

class1_data1 = np.random.multivariate_normal(class1_mean1, cov, 60)
class1_data2 = np.random.multivariate_normal(class1_mean2, cov, 60)
class2_data1 = np.random.multivariate_normal(class2_mean1, cov, 60)
class2_data2 = np.random.multivariate_normal(class2_mean2, cov, 60)

#Obtain 2-class dataset
all_data = np.vstack((class1_data1, class1_data2, class2_data1, class2_data2))
all_targets = np.array([1] * 60 + [-1] * 60 + [-1] * 60 + [1] * 60)
# all_targets = all_targets.reshape(all_targets.shape[0], 1)


#Create train test splits
n_total = all_data.shape[0]
indices = np.random.permutation(n_total)
n_train = int(0.6 * n_total)

train_idx = indices[:n_train]
test_idx = indices[n_train:]

tr_data = all_data[train_idx]
tr_targets = all_targets[train_idx]
te_data = all_data[test_idx]
te_targets = all_targets[test_idx]

# plt.scatter(te_data[te_targets == -1, 0], te_data[te_targets == -1, 1], color='cyan', label="Class -1", s=10)
# plt.scatter(te_data[te_targets == 1, 0], te_data[te_targets == 1, 1], color='pink', label="Class 1", s=10)
# plt.xlabel("Coordinate 1")
# plt.ylabel("Coordinate 2")
# plt.title("Testing Data")
# plt.legend()
# plt.grid()
# plt.show()

# endregion

#Classify this dataset using perceptron algorithm

def Perceptron(X_train, Y_train, X_test):
  # Inputs: Training data, Training labels, Testing data
  # Outputs: Testing data predictions, Weight vector representing the decision boundary

  epochs = 100  # Number of epochs
  # Append ones to the training data to account for the bias term
  X = np.hstack((X_train, np.ones((X_train.shape[0], 1))))
  # Initialize the weight vector (including bias) to zeros
  w = np.zeros(X.shape[1])
  
  # Learning rate (can be tuned)
  lr = 1.0
  
  for epoch in range(epochs):
    # Compute predictions for each sample
    predictions = (X @ w >= 0).astype(int)
    predictions = (predictions * 2) - 1
    # Update the weights for misclassified samples
    for i in range(X.shape[0]):
      error = Y_train[i] - predictions[i]
      if error != 0:
        # Update rule: w = w + lr * error * x_i
        w = w + lr * error * X[i]
              
  # Compute the test data predictions using the final weight vector
  X_test_bias = np.hstack((X_test, np.ones((X_test.shape[0], 1))))
  predictions = (X_test_bias @ w >= 0).astype(int)
  predictions = (predictions * 2) - 1
  
  return predictions, w

predictions, w = Perceptron(tr_data, tr_targets, te_data)

#Compute and print the accuracy
accuracy = np.mean(predictions == te_targets)
print("Accuracy with Perceptron:", accuracy)

# region: Decision boundary
x = np.linspace(-1000, 2000, 100)

plt.plot(x, predict_w3_minus1plus1(x, w), color='black', label = "Decision Boundary")
plt.scatter(te_data[te_targets == -1, 0], te_data[te_targets == -1, 1], color='cyan', label = "True Class -1", s=10)
plt.scatter(te_data[te_targets == 1, 0], te_data[te_targets == 1, 1], color='pink', label = "True Class 1", s=10)
plt.xlabel("Coordinate 1")
plt.ylabel("Coordinate 2")
plt.title("Perceptron classification for labels -1/1 (XOR Data)")
plt.grid()
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.legend()
plt.show()
# endregion

# endregion


```

<dt> <b> b. Implementation of Expectation and Maximization (EM) algorithm : </b> This part focuses on implementing the EM algorithm from scratch to estimate the parameters of the Gaussian Mixture Model (GMM). </dt>  

<dt>  1. Chose the number of mixtures (M) and initialize the parameters of GMM model  </dt>
<dd>  - Make sure that the diagonal elements of covariance are positive, and the weights of Gaussians are positive and sum to 1 for satisfying the axioms of probability.  </dd>

<dt>   2. Expectation step :  Evaluate the responsibility of Gaussians in generating the data points using the current estimated parameters i.e $\hat\mu_{k}$ , $\hat\Sigma_{k}$ and $\hat\pi_{k}$, here k indicates the Gaussian in GMM  </dt>
<br>

<dd>  <center> $\gamma_{nk} = \frac{\pi_{k} N ( \textbf{x}_{n} / \mu_{k} , \Sigma_{k} )}{ \sum_{m=1}^{M} \pi_{m} N ( x_{n} / \mu_{m} , \Sigma_{m} ) }    \;\;\;\;\;\; ∀ \;\;k=1,2,...,M \;\; and \;\; n=1,2,...,N$  </center>  </dd>

<br>

<dd>   Where N is the total number of data points and M is the total number of Gaussians in GMM   </dd>

<dd>   - $\gamma_{nk}$ represents the responsibility of $k^{th}$ Gaussian in generating the $n^{th}$ data point   </dd>

<dd>   - $\sum_{n=1}^{N}$ $\gamma_{nk}$ represents the the effective number of data points generated by $k^{th}$ Gaussian   </dd>


<dt>   3. Maximization step :  Update the parameters ($\hat\mu_{k}$ , $\hat\Sigma_{k}$ and $\hat\pi_{k}$) of GMM using the current resosibilities ($\gamma_{nk}$) by maximizing the likelihood. </dt>

<dd>
 - Effective number of data points generated by $k^{th}$ Gaussian : $N_{k}$ = $\sum_{n=1}^{N}$ $\gamma_{nk}$   
 </dd>

<dd>
 - Mean of $k^{th}$ Gaussian : $\hat{\mu_{k}} = \frac{1}{N_{k}} \sum_{n=1}^{N}\gamma_{nk} x_{n}$   
 </dd>

<dd>
 - Covariance of $k^{th}$ Gaussian : $\hat{\Sigma_{k}} = \frac{1}{N_{k}} \sum_{n=1}^{N} \gamma_{nk} ( x_{n} - \hat{\mu_{k}} )( x_{n} - \hat{\mu_{k}} )^{T}$   
 </dd>

<dd>
 - Weight of $k^{th}$ Gaussian : $\hat{\pi_{k}} = \frac{N_{k}}{N}$  
 </dd>

<dt>  4. Compute the log-likelihood with the updated parameters  </dt>
<dt>   5. Repeat the expectation and maximization steps untill convergence  </dt>

```python
#All imports
import numpy as np
import pandas as pd
import random as rand
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from scipy.stats import norm, multivariate_normal
import sys
import math

##########################################
#Part-6)-a: Generating synthetic data
##########################################

def generateData(pi, mu, sigma, N):
    ############################################################################
    #pi = weights of Gaussians, mu = mean of Gaussians, sigma = convariance of Gaussians, N = number of data points
    #Steps
    #1.Select th Gaussian following the probability distribution as pi
    #2.Sample the data points from the selected Gaussian
    #3.Repeat the process N times to get N data points
    #4.Return the sampled data
    ############################################################################
    #You may look into "random.choices" and "np.random.multivariate_normal" functions
    data = []
    labels = []
    random.seed(14)
    np.random.seed(14)
    for _ in range(N):
        # Select a Gaussian component (0 or 1) following the probability distribution in pi
        comp = rand.choices([0, 1], weights=pi, k=1)[0]
        # Sample a data point from the selected Bivariate Gaussian
        sample = np.random.multivariate_normal(mu[comp], sigma[comp])
        data.append(sample)
        labels.append(comp)

    ############################################################################
    return np.asarray(data), np.asarray(labels)
    ############################################################################


############################################################################
#Chose the appropriate means ($\mu$), covariances ($\Sigma$), and weights
############################################################################
mu1 = [-1.0, -1.0]   #Mean vector of component1 in GMM
sig1 = np.array([[0.15, 0.1], [0.1, 0.25]])  #Full covariance matrix of component1 in GMM
mu2 = [1.0, 1.0]   #Mean vector of component2 in GMM
sig2 = np.array([[0.3, -0.25], [-0.25, 0.25]])  #Full covariance matrix of component2 in GMM
pi = [0.6, 0.4]    #Prior probabilities


# Spherical covariances
sig1_s = np.array([[0.2, 0.0], [0.0, 0.2]])
sig2_s = np.array([[0.1, 0.0], [0.0, 0.1]])

# Diagonal covariances
sig1_d = np.array([[0.1, 0.0], [0.0, 0.2]])
sig2_d = np.array([[0.2, 0.0], [0.0, 0.1]])

############################################################################
#Create the synthetic data set
############################################################################

data_full, labels_full = generateData(pi, [mu1, mu2], [sig1, sig2], 200)
data_sp, labels_sp = generateData(pi, [mu1, mu2], [sig1_s, sig2_s], 200)
data_dg, labels_dg = generateData(pi, [mu1, mu2], [sig1_d, sig2_d], 200)

############################################################################
#Plot the synthetic data set
############################################################################

fig, ax = plt.subplots(1,3, figsize=(20,6))

custom_cmap = ListedColormap(['pink', 'cyan'])

# Visualize the data sampled from the GMM with full covariance matrix
ax[0].scatter(data_full[:, 0], data_full[:, 1], c=labels_full, cmap=custom_cmap, s=10)
ax[0].set_title('Full Covariance')
ax[0].set_xlim(-4, 4)
ax[0].set_ylim(-4, 4)

# Visualize the data sampled from the GMM with spherical covariance matrix
ax[1].scatter(data_sp[:, 0], data_sp[:, 1], c=labels_sp, cmap=custom_cmap, s=10)
ax[1].set_title('Spherical Covariance')
ax[1].set_xlim(-4, 4)
ax[1].set_ylim(-4, 4)

# Visualize the data sampled from the GMM with diagonal covariance matrix
ax[2].scatter(data_dg[:, 0], data_dg[:, 1], c=labels_dg, cmap=custom_cmap, s=10)
ax[2].set_title('Diagonal Covariance')
ax[2].set_xlim(-4, 4)
ax[2].set_ylim(-4, 4)

plt.show()
```

```python
# All imports
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal
from ipywidgets import interact, IntSlider
import ipywidgets as widgets

# Compute the log-likelihood of the data given the parameters
def likelihood(data, pi, mu, sigma, log_likelihood=False):
    N = data.shape[0]
    M = len(pi)
    
    likelihoods = np.zeros((N, M))
    for k in range(M):
        likelihoods[:, k] = pi[k] * multivariate_normal.pdf(data, mean=mu[k], cov=sigma[k])
    
    total_likelihood = np.sum(likelihoods, axis=1)
    
    if log_likelihood:
        return np.sum(np.log(total_likelihood))
    
    return likelihoods

# Expectation step: Compute responsibilities (gamma)
def expectation(data, pi, mu, sigma):
    N, M = data.shape[0], len(pi)
    gamma = np.zeros((N, M))
    
    likelihoods = likelihood(data, pi, mu, sigma)
    
    for k in range(M):
        gamma[:, k] = likelihoods[:, k] / np.sum(likelihoods, axis=1)
    
    return gamma

# Maximization step: Update parameters
def maximization(data, gamma):
    N, D = data.shape
    M = gamma.shape[1]

    Nk = np.sum(gamma, axis=0)
    
    # Update means
    mu = np.dot(gamma.T, data) / Nk[:, None]
    
    # Update covariance matrices
    sigma = np.zeros((M, D, D))
    for k in range(M):
        diff = data - mu[k]
        sigma[k] = np.dot(gamma[:, k] * diff.T, diff) / Nk[k]
    
    # Update weights
    pi = Nk / N

    return pi, mu, sigma

# Function to plot the contours of the estimated GMM distributions
def plot_contours(data, labels, pi, mu, sigma, gamma, iter_no=None):
    plt.figure(figsize=(6, 6))
    custom_cmap = ListedColormap(['pink', 'cyan'])
    # plt.scatter(data[:, 0], data[:, 1], c=labels, cmap=custom_cmap, alpha=1)
    assigned_component = np.argmax(gamma, axis=1)
    colors = np.array(['pink', 'cyan'])
    plt.scatter(data[:, 0], data[:, 1], c=colors[assigned_component], alpha=0.6, label='Data points')

    x, y = np.meshgrid(np.linspace(data[:, 0].min() - 1, data[:, 0].max() + 1, 100),
                       np.linspace(data[:, 1].min() - 1, data[:, 1].max() + 1, 100))
    xy = np.column_stack([x.ravel(), y.ravel()])
    
    for k in range(len(pi)):
        z = multivariate_normal.pdf(xy, mean=mu[k], cov=sigma[k])
        z = z.reshape(x.shape)
        plt.contour(x, y, z, levels=7, colors=['red', 'blue'][k % 2], alpha=0.6)
    
    plt.title(f"GMM Contours over Data Points (Iteration {iter_no})")
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.show()
```
